#### Lesson 1: Managing Missing Data (The NaN Problem)

```python
import pandas as pd
import numpy as np

data = {
    'Date': pd.to_datetime(['2026-01-01', '2026-01-02', '2026-01-03', '2026-01-04', '2026-01-05']),
    'Patient_ID': [101, 101, np.nan, 101, 101],
    'Heart_Rate': [72, np.nan, np.nan, 85, np.nan],
    'Notes': [np.nan, "Stable", np.nan, "Recovering", np.nan],
    'Empty_Col': [np.nan, np.nan, np.nan, np.nan, np.nan] # All missing
}

df = pd.DataFrame(data)
print("Original Messy Data:")
df
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.dropna()
```

```python
df.dropna(thresh=3)
```

```python
df.dropna(subset=['Patient_ID', 'Heart_Rate'])
```

```python
df.dropna(axis=1)
```

```python
df.dropna(axis=1, how="all")
```

```python
df
```

```python
df.fillna("Unknown")
```

```python
df.fillna({'Patient_ID': 101, 'Notes': "Unknown"})
```

```python
df
```

```python
df.fillna({'Patient_ID': 101, 'Notes': "Unknown"}, inplace=True)
```

```python
df
```

```python
df.ffill()
```

```python
df.bfill()
```

```python
df.ffill(limit=1)
```

#### Lesson 2: Data Type Integrity & Explicit Type Casting


```python
import pandas as pd
import numpy as np

# Simulate messy raw hospital data ingestion
messy_billing = {
    "patient_encounter_id": [9001, 9002, 9003, 9004],
    "admission_date": ["2026-05-10", "2026-05-11", "Malformed_Timestamp", "2026-05-13"],
    "copay_charge": [45.00, 150.75, "Missing_Record", 0.00],
    "encounter_class": ["ambulatory", "emergency", "emergency", "ambulatory"]
}

df = pd.DataFrame(messy_billing)
df
```

```python
df.dtypes
```

```python
df.memory_usage(deep=True).sum()
```

```python
# astype(), pd.to_numeric(), pd.to_datetime()
pd.to_numeric(df.copay_charge, errors="coerce", downcast="float")
```

```python
df.dtypes
```

```python
df.copay_charge = pd.to_numeric(df.copay_charge, errors="coerce", downcast="float")
df.dtypes
```

```python
df.admission_date = pd.to_datetime(df.admission_date, errors="coerce")
df.dtypes
```

```python
df.encounter_class = df.encounter_class.astype("category")
df.dtypes
```
